# vAIdya Case Lab on a free Colab GPU
Runs the full-precision **AyurParam** model (bharatgenai/AyurParam) on a T4 GPU and gives you a public link
to the app for about a week. Use this for live demonstrations; the Hugging Face Space runs a smaller
4-bit build on CPU.

**Before you start (one time):**
1. AyurParam is a gated model. Signed in to Hugging Face, open https://huggingface.co/bharatgenai/AyurParam and accept the access conditions.
2. Create a Read token at https://huggingface.co/settings/tokens.
3. In Colab, click the key icon (Secrets) on the left, add a secret named `HF_TOKEN` with the token as its value, and switch on *Notebook access*.
4. Runtime → Change runtime type → **T4 GPU** → Save. Then run each cell in order.

**Running on public URL: https://….gradio.live**, open that link. The status line at the top of the app turns to
*AyurParam ready* when the model has loaded.
