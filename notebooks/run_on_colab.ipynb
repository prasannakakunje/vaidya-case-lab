# vAIdya Case Lab on a free Colab GPU
Runs the full-precision **AyurParam** model (bharatgenai/AyurParam) on a T4 GPU and gives you a public link
to the app for about 72 hours. Use this for live demonstrations; the Hugging Face Space runs a smaller
4-bit build on CPU.

**Before you start:** Runtime → Change runtime type → **T4 GPU** → Save. Then run each cell in order.

```
!git clone -q https://github.com/prasannakakunje/vaidya-case-lab.git
%cd vaidya-case-lab
!pip -q install -r requirements-gpu.txt
```

Run the tests (should say *29 passed*) and the evaluation used in the report.

```
!pip -q install pytest
!python -m pytest -q
!python -m eval.run_eval | head -20
```

Start the app. The first start downloads AyurParam (about 11 GB, 3 to 6 minutes). When the cell prints
**Running on public URL: https://….gradio.live**, open that link. The status line at the top of the app turns to
*AyurParam ready* when the model has loaded.
